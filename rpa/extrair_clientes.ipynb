# Código-fonte RPA (Pyhton)

````python
import os
import requests
from bs4 import BeautifulSoup

URL_PAGINA = "https://digitalinnovationone.github.io/dio-lab-assistente-investimentos-rpa-n8n/"

# Cole aqui a URL copiada no nó Webhook do n8n.
# No Google Colab, NÃO use http://localhost:5678.
N8N_WEBHOOK = "COLE_AQUI_A_URL_PUBLICA_DO_WEBHOOK"

resposta_pagina = requests.get(URL_PAGINA, timeout=30)
resposta_pagina.raise_for_status()

soup = BeautifulSoup(resposta_pagina.text, "html.parser")
linhas = soup.select("#clientes tbody tr")

clientes = []
for linha in linhas:
    colunas = linha.find_all("td")
    if len(colunas) < 4:
        continue

    clientes.append({
        "nome": colunas[0].get_text(" ", strip=True),
        "email": colunas[1].get_text(" ", strip=True),
        "saldo": colunas[2].get_text(" ", strip=True),
        "perfil": colunas[3].get_text(" ", strip=True),
    })

print(f"Clientes encontrados: {len(clientes)}")
for cliente in clientes:
    print(cliente)

if N8N_WEBHOOK == "COLE_AQUI_A_URL_PUBLICA_DO_WEBHOOK":
    raise ValueError(
        "Cole a URL pública do Webhook n8n antes de executar."
    )

payload = {"clientes": clientes}

resposta_n8n = requests.post(
    N8N_WEBHOOK,
    json=payload,
    headers={"Content-Type": "application/json"},
    timeout=60,
)

print("Status HTTP:", resposta_n8n.status_code)
print("Resposta do n8n:")
print(resposta_n8n.text[:2000])
resposta_n8n.raise_for_status()
````
